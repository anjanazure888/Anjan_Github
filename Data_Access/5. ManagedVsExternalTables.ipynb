**Managed Table**

```python
spark.read.format("csv").option("header","true").load("abfss://demo@anjanltmstorage.dfs.core.windows.net/customers-100.csv").display()
```

```python
## Read data -- Rename columns to remove invalid characters and create a table customers

spark.read.format("csv").option("header","true").load("abfss://demo@anjanltmstorage.dfs.core.windows.net/customers-100.csv").withColumnRenamed("Customer Id","Customer_Id").withColumnRenamed("First Name","First_Name").withColumnRenamed("Last Name","Last_Name").withColumnRenamed("Phone 1","Phone_1").withColumnRenamed("Phone 2","Phone_2").withColumnRenamed("Subscription Date","Subscription_Date").write.mode("overwrite").saveAsTable("customers_Managed")
```

```python
## Read data to DF ; Rename all columns in one GO using FOR EACH Loop; Write data to Customers-1 Table
df=spark.read.format("csv").option("header","true").load("abfss://demo@anjanltmstorage.dfs.core.windows.net/customers-100.csv")
# Rename columns to remove invalid characters
for col in df.columns:
    df = df.withColumnRenamed(col, col.replace(' ', '_').replace('-', '_'))

## Create a table customers
df.write.mode("overwrite").saveAsTable("customers_Managed1") 
# df.display()
```

```python
%sql
Describe default.customers_Managed;
```

```python
%sql DESCRIBE EXTENDED default.customers_Managed
```

**External Table**

```python
df=spark.read.format("csv").option("header","true").load("abfss://demo@anjanltmstorage.dfs.core.windows.net/customers-100.csv").withColumnRenamed("Customer Id","Customer_Id").withColumnRenamed("First Name","First_Name").withColumnRenamed("Last Name","Last_Name").withColumnRenamed("Phone 1","Phone_1").withColumnRenamed("Phone 2","Phone_2").withColumnRenamed("Subscription Date","Subscription_Date")

path = "abfss://demo@anjanltmstorage.dfs.core.windows.net/Table/Customers_External"

df.write.mode("overwrite").format("delta").save(path)

```

```python
%sql
create table default.customers_external1
using parquet
location "abfss://demo@anjanltmstorage.dfs.core.windows.net/Table/Customers_External/"
```
