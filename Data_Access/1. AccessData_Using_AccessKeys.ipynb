Access data from ADLSGen2 through Access Keys
1.  Set Spark Configuration
2.  Use Abfss Driver to refer the data
3.  Display the Data 

```python
# set the configuration
spark.conf.set("fs.azure.account.key.anjanltmstorage.dfs.core.windows.net","Anjan_GetACCESSKEY")
```

```python
# just refer the data from the ADLS gen2 
dbutils.fs.ls("abfss://demo@anjanltmstorage.dfs.core.windows.net")
```

```python
display(dbutils.fs.ls("abfss://demo@anjanltmstorage.dfs.core.windows.net"))
```

Read Data from ADLS storage file

```python
# just read the data and displays Columns and datatypes
spark.read.csv("abfss://demo@anjanltmstorage.dfs.core.windows.net/customers-100.csv")
```

```python
# dataframe is used to store the read data and display shows the data with Columns C1,C2 ....
df=spark.read.csv("abfss://demo@anjanltmstorage.dfs.core.windows.net/customers-100.csv")    
df.show() ## not readable format
```

```python
#DF - Data frame to store read data with Header option 
df = spark.read.option("header", "true").csv("abfss://demo@anjanltmstorage.dfs.core.windows.net/customers-100.csv")
df.display()  ## well readable format
```
