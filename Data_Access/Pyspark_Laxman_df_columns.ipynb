```python
from pyspark.sql.types import StructType,StructField,StringType, IntegerType, TimestampType

schema=StructType([

    StructField("Id",IntegerType(),False)
    ,StructField("FName",StringType(),True)
    ,StructField("LName",StringType(),True)
    ,StructField("Age",IntegerType(),True)
])
data=[
    (1,"John","raj",20),
    (2,None,None,30),
    (3,"Mike",None,40),
    (4,None,"sharma",39)
    ,(5,None,"patel",31)
    ,(6,"raj","raj",31)

]

dff= spark.createDataFrame(data,schema)
dff.display()
```

```python
dff.where(dff.FName.isNull()).display()
```

```python
dff.where(dff.LName.isNotNull()).display()
```

```python
dff.where(dff.FName==dff.Lname).display() ## null is not equal to null
```

```python
dff.where(dff.FName.eqNullSafe(dff.LName)).display()
```
