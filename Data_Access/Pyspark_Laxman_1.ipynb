**LIST** : List is a collection of Elements separated by Commas,enclosed by [ ] square braces

```python
L1=["anji","kusuma",40]
L2=["Balu","Neelima",41]
# data=(L1,L2)
data=[L1,L2]
df_list=spark.createDataFrame(data,schema=('name string,wifename string,age int'))
df_list.display()
df_list.show()
df_list.printSchema()
print(type(df_list))
df_list.schema
```

**Tuple** : List is a collection of Elements separated by Commas,enclosed by ( ) open/close braces

```python
T1=("raji","venu",40)
T2=("madhuri","Prasad",41)
# data=(L1,L2)
data=[T1,T2]
df_tuple=spark.createDataFrame(data,schema=('name string,wifename string,age int'))
df_tuple.display()
df_tuple.show()
df_tuple.printSchema()
print(type(df_tuple))
df_tuple.schema
```

**Dictionary** : key value pairs separated by commas ,enclosed by { } flower braces

**CREATE DATAFRAME IN SPARK**

```python
#Syntax : spark.createDataFrame(data,schema=None)
#data : list of lists or list of tuples or pandas.DataFrame or spark.DataFrame or pyspark.sql.Row or iterator of pyspark.sql.Row
#schema : StructType or list of str or str or None
#schema is optional and can be inferred from data or specified as a list of column names or a DDL-formatted string.
#schema is a StructType object which is a list of StructField objects.
#StructField is a class that represents a field in a StructType.
#StructField has three attributes: name, dataType, and nullable.
#name is a string that represents the name of the field.
#dataType is a DataType object that represents the data type of the field.
#nullable is a boolean that represents whether the field can be null.
#nullable is optional and defaults to True.
#StructType is a class that represents a schema.
#StructType has one attribute: fields.
#fields is a list of StructField objects.
#StructType is a subclass of DataType.
#DataType is a class that represents a data type.
#DataType has one attribute: typeName.
#typeName is a string that represents the name of the data type.
#StructType is a subclass of DataType.
#StructType has one attribute: fields.
#fields is a list of StructField objects.
#StructField is a class that represents a field in a StructType.
#StructField has three attributes: name, dataType, and nullable.
#name is a string that represents the name of   the field.
#dataType is a DataType object that represents the data type of the field.
#nullable is a boolean that represents whether the field can be null.
#)
```

**StructType and StructField**

```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,ArrayType,MapType
Data=[
         (1,"Rajesh","Khanna",40,[100,150,300],{"gender":"male"},{"City":"Mumbai","state":"Maharashtra"})
        ,(2,"Rasi","Khanna",41,[90,144,320],{"gender":"male"},{"City":"TPG","state":"AP"})
        ,(3,"Abhi","Khanna",39,[110,155,310],{"gender":"male"},{"City":"Hyd","state":"TG"})
     ]
    
Schema = StructType([
        StructField("Id", IntegerType(), True),
        StructField("FN", StringType(), True),
        StructField("LN", StringType(), True),
        StructField("Age", IntegerType(), True),
        StructField("marks", ArrayType(IntegerType()), True),
        StructField("gender", MapType(StringType(), StringType()), True),
        StructField("address", MapType(StringType(), StringType()), True)
    ])

df=spark.createDataFrame(Data,Schema)
df.display()
df.printSchema()
df.schema
```
